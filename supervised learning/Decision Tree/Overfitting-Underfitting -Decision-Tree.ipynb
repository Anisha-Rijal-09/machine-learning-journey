```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
```

```python
 df =pd.read_csv("Social_Network_Ads.csv")
df
```

```python
def analyzer(max_depth):
    df =pd.read_csv("Social_Network_Ads.csv")
    X= df.iloc[:,2:4]
    y=df.iloc[:,-1]
    clf = DecisionTreeClassifier(max_depth=max_depth)
    clf.fit(X,y)
    a=np.arange(start=X.iloc[:,0].min()-1, stop=X.iloc[:,0].max()+1,step=0.5)
    b=np.arange(start=X.iloc[:,1].min()-1000, stop=X.iloc[:,1].max()+1000,step=500)


    XX,yy=np.meshgrid(a,b)
    input_array=np.array([XX.ravel(),yy.ravel()]).T
    labels=clf.predict(input_array)
    plt.contourf(XX,yy,labels.reshape(XX.shape),alpha=0.5)
    plt.scatter(X.iloc[:,0],X.iloc[:,1],c=y)
     

        


    
```

```python
analyzer(max_depth=None) #overfitting
```

```python
analyzer(max_depth=1) #underfitting
```

```python

analyzer(max_depth=5)
```
